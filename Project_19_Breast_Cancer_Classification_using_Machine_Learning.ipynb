```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data=pd.read_csv('/content/data.csv')
```

```python
data['diagnosis']=data['diagnosis'].map({'B':1 , 'M':0})
```

```python
data.groupby('diagnosis').mean()
```

```python
X = data.drop(columns=['id' , 'diagnosis', 'Unnamed: 32' ], axis=1)
Y = data['diagnosis']
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size=0.2 , random_state=42)
```

```python
print(X_train.shape , X.shape, X_test.shape)
```

```python
model=LogisticRegression()
```

```python
X_train.isnull().sum()
```

```python
model.fit(X_train , Y_train)
```

```python
prediction_train=model.predict(X_train)
accuracy_train_score=accuracy_score(Y_train , prediction_train)
```

```python
print(accuracy_train_score)
```

```python
prediction_test=model.predict(X_test)
accuracy_test_score=accuracy_score(Y_test , prediction_test)
```

```python
print(accuracy_test_score)
```
